## 📊 分析概要

このレポートは、福岡市中央区（天神エリア）における商業空間の変質を、2つの独立した統計分析から科学的に証明するものです。

### 🎯 主仮説
**天神エリアは「外部からの資本流入」と「個人商店の消滅」によって、個性的な商業空間から均質化された「東京化」した空間へと転換している**

### 📈 2つの証拠
1. **流入依存度分析** (notebook/01_population_dynamics_api.ipynb)

   - データ源：e-Stat API（事業所統計）

   - 意味：「流入者が出会う商業環境がどう変わったか」を示す
