## セクション1: 映像構成案

### 🎬 5段階の映像構成

**【第1段階】問題提起（0:00-1:00）**

- 映像：古い天神の写真・映像素材
- グラフ：テナント構成比較（個人経営率の低さを強調）

**【第2段階】統計的根拠 - 人口動態（1:00-2:30）**
- ナレーション：「しかし今、天神に流入する人々の正体が変わった」
- グラフ：`chuo_inflow_dependency.png`
- ポイント：流入依存度の推移グラフで「外部資本の支配」を可視化

**【第3段階】統計的根拠 - テナント構成（2:30-4:00）**
- ナレーション：「流入してくる消費者の増加に応じて、商業空間も変わった」
- グラフ：`tenant_organization_composition.png`（円グラフ）
- グラフ：`tenant_organization_breakdown.png`（詳細構成）

**【第4段階】統合分析（4:00-5:00）**
- ナレーション：「外部からの資本流入と、個人商店の消滅」
- 「この2つの流れが交差するとき、天神は変わった」
- ビジュアル：2つのグラフを並べて表示
- ポイント：因果関係と相乗効果を示唆

**【第5段階】結論（5:00-）**
- ナレーション：「これが『東京化』の姿である」
- 映像：現在の天神のチェーン店映像
- テロップ：主要統計数値の表示

## セクション2: グラフ使用ガイド

### 📊 生成グラフと用途の対応表

| ファイル | 内容 | 映像内での用途 | タイミング |
|---------|------|-------------|----------|
| `chuo_inflow_dependency.png` | 流入依存度の月別推移＋社会動態増減 | 「市外からの流入が9割超」を可視化 | 第2段階（1:00-2:30） |
| `tenant_organization_composition.png` | 円グラフ + 5区の比較棒グラフ | 「個人経営6.8% vs 法人93.2%」の強調 | 第3段階前半（2:30-3:15） |
| `tenant_organization_breakdown.png` | 経営組織別詳細構成（円グラフ＋横棒） | 「法人内訳の詳細」説明 | 第3段階後半（3:15-4:00） |

### 🎨 グラフの編集・活用方法

#### グラフ1の活用
```
上グラフ（流入依存度）：
  → 右肩上がりの傾向を強調
  → テロップ：「令和6年1月～令和7年12月」
  → テロップ：「流入依存度の推移」

下グラフ（社会動態増減）：
  → 正負の変動を背景にしながら
  → 赤線（流入依存度）が常に高い水準に保たれていることを強調
  → ナレーション：「数百人の人口増減より、その構成が重要」
```

#### グラフ2の活用
```
左円グラフ：
  → パイの大きさを視覚化（個人vs法人の圧倒的差）
  → 数字を大きく表示：「6.8%」「93.2%」
  → テロップ：「福岡市中央区 2024年」

右棒グラフ：
  → 5区を横並びに比較
  → 中央区と博多区の低さを色分け（強調）
  → テロップ：「他の区との比較」
```

#### グラフ3の活用
```
左円グラフ：
  → 「会社」が84%を占めることを強調
  → テロップ：「個人は1割以下」

右横棒グラフ：
  → 「会社」の圧倒的な数を視覚化
  → スクロール表示で、棒の長さを見守らせる
  → ナレーション：「13,000件を超える会社形態の事業所」
```

## セクション3: ナレーション案

### 段階別ナレーション案（素案）

**【第1段階：問題提起】**
```
「1980年代、2000年代の天神は、個性的な商店が立ち並ぶ商業地だった。
 親から子へ、代々受け継がれた小さな店舗が、街の個性を作っていた。
 だが、今の天神の顔は変わった。」
```

**【第2段階：人口動態の視点】**
```
「統計が示すのは、天神に流入する人々の正体の変化である。
 令和6年から令和7年にかけて、天神の新規転入者の9割以上が
 市外からの流入であった。
 
 つまり、地元で生まれ育った人々ではなく、
 仕事や消費を目的に市外から流入してくる人々が、
 天神の経済を支えているということである。」
```

**【第3段階：テナント構成の視点】**
```
「この変化を象徴するのが、テナント構成の劇的な変容である。
 
 2024年の中央区の事業所のうち、
 個人商店は、わずか6.8%に過ぎない。
 残りの93%以上が、法人企業による店舗である。
 
 しかも、その大半は『会社』という形態の企業。
 つまり、大手チェーン、フランチャイズ、
 グローバル資本傘下の店舗である。」
```

**【第4段階：統合分析】**
```
「外部からの流入＋商業空間の法人化。
 この二つの現象が同時に進行するとき、
 都市は大きく変わる。
 
 個性的で、多様で、地に根ざした商業空間から、
 どこの都市でも見られる、
 グローバル企業に最適化した『均質な商業空間』へ。
 
 これが『東京化』と呼ばれる現象である。」
```

**【第5段階：結論】**
```
「統計は、かつての『天神らしさ』が、
 数字に変わった世界の中で消えゆくことを物語っている。
 
 これは衰退ではなく、別の形の発展かもしれない。
 だが、その代償として、地域に根ざした文化は
 確実に失われつつあるのだ。」
```

## セクション5: チェックリスト

### 制作前の確認事項

```
□ グラフの高解像度化（300 DPI が確保されているか）
□ グラフの背景色が映像背景に合致しているか（dark_background確認）
□ 日本語フォントが正しく表示されるか（テロップ作成後の確認）
□ タイミング：各グラフの出現時間を秒単位で決定
□ ナレーションの語呂合わせ：映像とナレーションのシンク確認
□ 古い天神の写真素材の確保：オープニング映像用
□ 現在の天神のチェーン店映像：エンディング映像用
```

### グラフ配置案

```
【タイムライン】
0:00-1:00   → 問題提起（古い映像＋テロップ）
1:00-2:30   → グラフ1（chuo_inflow_dependency.png）
2:30-3:30   → グラフ2左（tenant_organization_composition円グラフ）
3:30-4:00   → グラフ2右（区別比較棒グラフ）＋グラフ3
4:00-5:00   → グラフ1＆グラフ2の並列表示（統合分析）
5:00-       → 現在の天神映像＋最終テロップ
```
