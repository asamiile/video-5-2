### 🎨 グラフの編集・活用方法

#### グラフ1の活用
```
上グラフ（流入依存度）：
  → 右肩上がりの傾向を強調
  → テロップ：「令和6年1月～令和7年12月」
  → テロップ：「流入依存度の推移」

下グラフ（社会動態増減）：
  → 正負の変動を背景にしながら
  → 赤線（流入依存度）が常に高い水準に保たれていることを強調
  → ナレーション：「数百人の人口増減より、その構成が重要」
```

#### グラフ2の活用
```
左円グラフ：
  → パイの大きさを視覚化（個人vs法人の圧倒的差）
  → 数字を大きく表示：「6.8%」「93.2%」
  → テロップ：「福岡市中央区 2024年」

右棒グラフ：
  → 5区を横並びに比較
  → 中央区と博多区の低さを色分け（強調）
  → テロップ：「他の区との比較」
```

#### グラフ3の活用
```
左円グラフ：
  → 「会社」が84%を占めることを強調
  → テロップ：「個人は1割以下」

右横棒グラフ：
  → 「会社」の圧倒的な数を視覚化
  → スクロール表示で、棒の長さを見守らせる
  → ナレーション：「13,000件を超える会社形態の事業所」
```

#### グラフ4【NEW】の活用：land_price_individual_shops_summary_2024.png
```
左パネル（商業地価）：
  → 「7,010,000 円/㎡」という数字の大きさを強調
  → テロップ：「福岡市中央区の商業地平均公示地価」
  → テロップ：「小規模事業者には負担不可能な水準」
  → ナレーション：「小さな商店が払える賃料ではない」

右パネル（個人/法人比率）：
  → 6.8% の小ささを視覚化（わずかな切れ込み）
  → 93.2% の大きさを対比
  → テロップ：「個人経営：6.8% → 法人経営：93.2%」
  → ナレーション：「その結果が、ここに表れている」

【映像演出アイデア】
  • 左パネルから右パネルへのズームアウト
  • テロップで「地価↑ → 個人経営↓」の因果関係を矢印で表示
  • ナレーション：「高い地価が、個人商店を消す」
```

## セクション3: 分析ノートブックのデータサマリー

### 分析01：人口動態API

**分析対象**：福岡市中央区（令和6年1月～令和7年12月）

**出力結果**
- グラフファイル: `chuo_inflow_dependency.png`
- グラフ1: 中央区の市外流入比率の月別推移
- グラフ2: 社会動態増減と流入依存度の関係性（二軸グラフ）

**データ解釈上の重要点**
- **流入依存度が高い** = 市外からの転入者が新規転入に占める割合が大きい
- **社会動態増減** = 転入者数から転出者数を差し引いた値（プラス=流入超過）
- **ナレーション活用**: 「数百人の人口増減より、その構成が重要」

---

### 分析02：流動性データCSV

**分析対象**：中央第１（天神・大名）、南第１（住宅街）ほか（2010年～2024年）

**出力結果**
- グラフファイル: `graph_turnover_comparison.png`、`graph_chuo1_inflow_outflow.png`
- グラフ1: 中央第１と南第１の流動性推移比較（2015年再開発開始時点表示）
- グラフ2: 中央第１の転入数と転出数の二軸グラフ

**【重要】街の硬直化指標**
- **のべ移動数の実態**: 2010年 4,995人 → 2024年 4,647人（-7.0%減少）
- **解釈**: 地元住民の世代交代機能が低下 = 街の代謝が落ちている
- **ナレーション活用**: 「数百人の人口増減より、その構成が重要。絶対数すら減っている」

**データ解釈上の注意**
- 絶対数の変化: 街の総入れ替わり人数は実は減少している（2010⇒2024）
- 構成比の変化: しかし流入依存度（市外からの比率）は上昇している
- 意味: 「人口入れ替わりの密度は落ちているのに、外部資本層による上書きの密度は上がっている」

## セクション3: ナレーション案

### 段階別ナレーション案（素案）

**【第1段階：問題提起】**
```
「1980年代、2000年代の天神は、個性的な商店が立ち並ぶ商業地だった。
 親から子へ、代々受け継がれた小さな店舗が、街の個性を作っていた。
 だが、今の天神の顔は変わった。」
```

**【第2段階：人口動態の視点】**
```
「統計が示すのは、天神に流入する人々の正体の変化である。
 令和6年から令和7年にかけて、天神の新規転入者の9割以上が
 市外からの流入であった。
 
 つまり、地元で生まれ育った人々ではなく、
 仕事や消費を目的に市外から流入してくる人々が、
 天神の経済を支えているということである。」
```

**【第3段階：テナント構成の視点】**
```
「この変化を象徴するのが、テナント構成の劇的な変容である。
 
 2024年の中央区の事業所のうち、
 個人商店は、わずか6.8%に過ぎない。
 残りの93%以上が、法人企業による店舗である。
 
 しかも、その大半は『会社』という形態の企業。
 つまり、大手チェーン、フランチャイズ、
 グローバル資本傘下の店舗である。」
```

**【第3.5段階：商業地価分析【NEW】】**
```
「では、なぜ個人商店は消えたのか。
 その答えが、商業地の急騰する地価にあります。
 
 福岡市中央区の商業地は、
 一平方メートルあたり700万円を超えるエリアもあります。
 
 小さな個人商店が払えるテナント料ではありません。
 
 相続税、固定資産税、営業を続けるための賃料……
 すべてが、高い地価に基づいて計算されます。
 
 低い利益率で運営してきた個人商店は、
 この経済的障壁に対応することができません。
 
 結果として、個人経営者は選択肢を奪われ、
 街から消えてゆくのです。」
```

**【第4段階：統合分析】**
```
「外部からの流入、地価の上昇、そして個人商店の消滅。
 この三つの流れが同時に進行するとき、
 都市は劇的に変わります。
 
 個性的で、多様で、地に根ざした商業空間から、
 どこの都市でも見られる、
 グローバル企業に最適化した『均質な商業空間』へ。
 
 これが『東京化』と呼ばれる現象である。
 そして天神は、その典型的な事例なのです。」
```

**【第5段階：結論】**
```
「統計は、かつての『天神らしさ』が、
 数字に変わった世界の中で消えゆくことを物語っている。
 
 これは衰退ではなく、別の形の発展かもしれない。
 だが、その代償として、地域に根ざした文化は
 確実に失われつつあるのだ。
 
 街が変わるとき、何が守られ、何が失われるのか。
 その問いを、データは静かに突きつけている。」
```
